```python
from langgraph.graph import StateGraph, START, END

#from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
#from langchain_opneai import ChatOpenAI

from langchain_ollama import ChatOllama
from typing import TypedDict
from dotenv import load_dotenv
import os
```

```python
# Reuse the existing `model`; no ChatOllama import or reassignment is needed.

model = ChatOllama(
    model="llama3.2",
    temperature=0.7
)
```

```python
#define a state
class BlogState(TypedDict):
    title:str
    outline:str
    content:str
```

```python
def create_outline(state:BlogState)->BlogState:
    #fetch title
    title= state['title']
    #call llm generate outline
    prompt = f'Generate the blog on the topic-{title}'
    outline = model.invoke(prompt).content
    #update the state
    state['outline'] = outline
    return state
```

```python

def create_blog(state:BlogState)->BlogState:
    title = state['title']
    outline = state['outline']
    prompt = f'write a detailed title on topic {title} using the following outlilne \n {outline}'
    content = model.invoke(prompt).content
    state['content'] = content
    return state
```

```python
#create a graph
graph= StateGraph(BlogState)

#create node
graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)

graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog', END)

workflow = graph.compile()
```

```python
#execution

initial_state= {'title': 'rise a ai in india'}
final_state= workflow.invoke(initial_state)
print(final_state)
```
