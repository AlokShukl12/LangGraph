```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class CriScore(TypedDict):
    runs:int
    balls:int
    four:int
    six:int

    sr : float
    bpb:float
    boundery_percent:float
    summery:str
```

```python
def calculate_sr(state:CriScore):
    sr = (state['runs']/state['balls'])*100
    return {'sr':sr}
```

```python
def calculate_bpb(state:CriScore):
    bpb = state['balls']/(state['four']+state['six'])
    return {'bpb':bpb}
```

```python
def calculate_boundery_percent(state:CriScore):
    boundery_percent =(((state['four']*4)+(state['six']*6))/state['runs'])*100
    return {'boundery_percent' : boundery_percent}
```

```python
def summery(state:CriScore):
    summery=f"""
Strike-rate - {state['sr']} /n
Ball per bounery - {state['bpb']} /n 
Boundery percent - {state['boundery_percent']} /n 
""" 
    return {'summery':summery}
```

```python
graph = StateGraph(CriScore)

graph.add_node('calculate_sr',calculate_sr)
graph.add_node('calculate_bpb',calculate_bpb)
graph.add_node('calculate_boundery_percent',calculate_boundery_percent)
graph.add_node('summery',summery)

graph.add_edge(START,'calculate_sr')
graph.add_edge(START,'calculate_bpb')
graph.add_edge(START,'calculate_boundery_percent')

graph.add_edge('calculate_sr','summery')
graph.add_edge('calculate_bpb','summery')
graph.add_edge('calculate_boundery_percent','summery')

graph.add_edge('summery',END)

workflow = graph.compile()
```

```python
initial_state = {
    'runs':100,
    'balls': 50,
    'four':6,
    'six':4,
}
final_state= workflow.invoke(initial_state)
print(final_state)
```
